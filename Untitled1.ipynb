```python
# swapping of two numbers using a third variable

a = int(input("Enter a number a: "))
b = int(input("Enter a number b: "))
c = a
a = b
b = c

print("a: ", a)
print("b: ", b)
```

```python
# swapping of two numbers not using an extra variable

a = int(input("Enter a number a: "))
b = int(input("Enter a number b: "))
a = a+b
b = a-b
a = a-b
print("a: ", a)
print("b: ", b)
```

```python
# swapping of two numbers

a = int(input("Enter a number a: "))
b = int(input("Enter a number b: "))
a,b = b,a
print("a: ", a)
print("b: ", b)
```
